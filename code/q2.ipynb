```python
citiesy1 = df_years[df_years["year"] == config.Y1]["city"].unique()
citiesy2 = df_years[df_years["year"] == config.Y2]["city"].unique()

print(f"Cities in {config.Y1}:", len(citiesy1))
print(f"Cities in {config.Y2}:", len(citiesy2))

new_cities = set(citiesy2) - set(citiesy1)
print(new_cities)
print(f"New cities in {config.Y2}:", len(new_cities))
```

## Check assumptions and test whether there is a difference

### Are the data Parametric or Non-parametric?
I used the Shapiro Wilk test to determine normality.

* $H_0$: population is normally distributed.
* $H_A$: population is not normally distributed.


Both $p$ values are very small (<0.05), therefore reject $H_0$. Data is not normally distributed.

```python
df_years = df_years[['year', 'pm25_concentration']].dropna()

pm25_y1 = df_years[df_years['year'] == config.Y1]['pm25_concentration']
pm25_y2 = df_years[df_years['year'] == config.Y2]['pm25_concentration']

shapiro_y1 = shapiro(pm25_y1)
shapiro_y2 = shapiro(pm25_y2)

print(f"Shapiro-Wilk test for {config.Y1}: W={shapiro_y1.statistic}, p-value={shapiro_y1.pvalue}")
print(f"Shapiro-Wilk test for {config.Y2}: W={shapiro_y2.statistic}, p-value={shapiro_y2.pvalue}")
```

### Wilcoxon Rank Sum Test


* $H_0$: $\mu_1 = \mu_2$
* $H_A$: $\mu_1 \neq \mu_2$

The $p$ value is $<0.05$. We reject $H_0$. There is a difference between the two years. 


```python
mannwhitney_result = mannwhitneyu(pm25_y1, pm25_y2, alternative='two-sided')

print(f"Mann Whitney test result: W={mannwhitney_result.statistic}, p-value={mannwhitney_result.pvalue}")
```
